# Creating random tensors

```python
# Create random tensor of size (3, 5) => 15 elements

3 * 5
```

```python
random_tensor = torch.rand(3, 5)
random_tensor
```

```python
random_tensor.ndim
```

```python
random_tensor.shape
```

Creating a random tensor with similar shape to an image

```python
# rand(height, with, number of color channels)

random_image_size_tensor = torch.rand(224, 224, 3)
random_image_size_tensor
```

```python
random_image_size_tensor.shape
```

```python
random_image_size_tensor.ndim
```

Creating tensors with zeros and ones

```python
# Create a tensorof all zeros
zeros = torch.zeros(size=(3,5))
print(zeros)
```

```python
zeros * random_tensor
```

```python
# Create a tensor of all ones
ones = torch.ones(size=(3, 4))
ones
```

```python
# Check tensor datatype
ones.dtype
```

```python
# Check tensor datatype
random_image_size_tensor.dtype
```

Creating a range of tensors and tensors-like

```python
# Using torch.range()
# If you get a deprecated message, use torch.arange()
# Create tensor from 1 to 10
one_to_ten = torch.arange(start=1, end=11, step=1)
one_to_ten
```

```python
# Create tensors-like
ten_zeros = torch.zeros_like(input=one_to_ten)
ten_zeros
```

Tensor datatypes

```python
# Float 32 tensor
float_32_tensor = torch.tensor([3.0, 6.0, 9.0], dtype=None)
float_32_tensor
```

```python
float_32_tensor.dtype
```
